# 1. What Is a Sequence in an RNN?

A **sequence** is an ordered collection of data points. The order matters because each item may relate to items that came before or after it.

Unlike independent records, sequence data often contains information that depends on context or time.

## Examples

### A sentence (text sequence)

**I → love → AI → technology**

Changing the word order can change the sentence's meaning.

### Stock prices (time sequence)

| Day | Price |
|:---|---:|
| Day 1 | ₹100 |
| Day 2 | ₹105 |
| Day 3 | ₹102 |
| Day 4 | ₹110 |

Each price is associated with a particular point in time.

Other examples of sequence data include:

- Audio and speech signals
- Weather measurements over time
- Daily sales records
- Video frames
- Sensor readings

## 2. How does RNN process a sequence?

A Recurrent Neural Network (RNN) is a neural network designed to process sequential data. It processes the sequence one element at a time and maintains a hidden state, which carries information from previous steps.

### Example: Processing a sentence

Consider the sentence: “I love machine learning.” The RNN processes its words in order:

1. **I** — Reads the first word.
2. **love** — Uses the current word and the previous hidden state.
3. **machine** — Carries forward information from earlier words.
4. **learning** — Updates the hidden state using the current word.

At every step, the network updates its hidden state based on the current input and the previous hidden state.

The image shows the same RNN in two views. On the left, the loop represents a recurrent connection: the hidden state is fed back into the network for the next step. On the right, the network is **unfolded** across time to show how it processes each item in a sequence.

### What the parts represent

- **Input (`x_t`)**: The item being processed at time step `t`. In the sentence example, each input represents one word.
- **Hidden state (`h_t`)**: The network's running memory. It combines the current input with information passed forward from the previous time step.
- **Input-to-hidden weights (`U`)**: Transform the current input before it contributes to the hidden state.
- **Hidden-to-hidden weights (`V`)**: Transform the previous hidden state and pass its information to the next step. This is the recurrent connection.
- **Hidden-to-output weights (`W`)**: Map a hidden state to an output at that time step. The diagram labels these outputs `L_t`; output notation varies by diagram, and a loss is calculated only when predictions are compared with targets.

At each time step, the hidden state is updated using the current input and previous hidden state:

`h_t = f(U x_t + V h_(t-1) + b_h)`

The output is calculated from the hidden state:

`y_t = g(W h_t + b_y)`

Here, `f` and `g` are activation functions, and `b_h` and `b_y` are biases. The same `U`, `V`, and `W` weights are reused at every time step. This lets the RNN process sequences of different lengths while carrying context forward from one step to the next.
